**Traccia:** Crea un prompt che estragga i dettagli chiave (nome, età, professione) da un testo descrittivo su una persona.

**Obiettivo:** Estrarre informazioni strutturate da un testo libero.

```python
# TO DO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-5-mini",
    messages=[
        {"role" : "system", 
         "content" : """
            Sei un sistema di estrazione dati. 
            Il tuo compito è analizzare il testo fornito ed estrarre esclusivamente tre informazioni sulla persona: 
            i nominativi (indicando nome e, se presente, il cognome), 
            l'età (espressa solamente in numero, es. 38)
            e la professione (indicando solo il nome del mestiere senza aggiungere dati ulteriori o inventarne, es. Dottore). 
            Nello svolgimento del compito devi basarti rigorosamente solo sul testo fornito, 
            senza creare o cercare altre informazioni esterne, 
            se uno di questi dati non è presente nel brano devi scrivere NON SPECIFICATO.
            """},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```

Il mio prompt: 

Hidetaka Miyazaki è un celebre autore di videogiochi e game director giapponese di 52 anni, noto in tutto il mondo per essere il presidente di FromSoftware e l'ideatore della saga di Dark Souls. Entrato tardi nell'industria videoludica dopo una laurea in scienze sociali, ha rivoluzionato il medium creando il genere dei soulslike, caratterizzato da una sfida impegnativa, mondi dark fantasy interconnessi e una narrazione frammentata ispirata alle sue letture giovanili. Oltre alla trilogia di Dark Souls, ha diretto capolavori acclamati dalla critica come Bloodborne, Sekiro: Shadows Die Twice ed Elden Ring

Output:

Nominativi: Hidetaka Miyazaki
Età: 52
Professione: Autore di videogiochi
