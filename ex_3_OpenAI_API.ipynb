```python
# TO DO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role" : "system", 
         "content" : """
            Sei un giornalista esperto nelle sintesi di articoli.
            Il tuo compito è riassumere l'articolo di notizie fornito, estraendo esclusivamente i punti essenziali e i fatti principali della vicenda. 
            Scrivi il riassunto in un unico paragrafo completo, evitando di andare a capo o di inserire titoli e commenti personali. 
            La lunghezza totale del tuo output non deve superare le 30-35 parole e in alcun modo i 255 caratteri, spazi e punteggiatura inclusi.
            """},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```
