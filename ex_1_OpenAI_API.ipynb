**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python
# TO DO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-4o-mini", #modificato da gpt-3.5-turbo poichè deprecato
    messages=[
        {"role" : "system", 
         "content" : """
            Agisci come un traduttore professionista. I tuoi compiti sono i seguenti: 
            Traduci un testo dall Italiano all Inglese se il testo è in Italiano e viceversa. 
            Restituisci solo il testo tradotto, senza commenti o introduzioni.
            """},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```

Il mio prompt:

For those who come after.

Output:

Per coloro che verranno dopo.
