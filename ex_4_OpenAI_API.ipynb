```python
# TO DO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")
data = {"name": "Giovanni", "age": 30, "city": "Roma", "profession": "Ingegnere"}


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role" : "system", 
         "content" : """
            Sei uno scrittore esperto. Utilizzando la lista di dati strutturati Python, 
            elabora un testo descrittivo e fluido. Utilizza solo i dati come imput. 
            """},
        {"role": "user", "content": str(data)}
    ],
)

print(complention.choices[0].message.content)
```

Output:
