```python
# TO DO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role" : "system", 
         "content" : """
            Sei un insegnante di analisi del testo.
            Analizza attentamente il breve brano fornito e genera 5 domande educative in testo puro per verificare che il lettore abbia colto i concetti principali. 
            Formula domande chiare, stimolanti e adatte a un contesto didattico, basandoti esclusivamente sulle informazioni presenti nel testo senza richiedere conoscenze esterne, 
            e riporta subito sotto ciascuna domanda la relativa risposta corretta per facilitare la verifica.
            """},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```
